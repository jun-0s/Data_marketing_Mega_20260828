```python
import requests
from bs4 import BeautifulSoup

url = "https://news.daum.net/tech"

headers = {
    "user-agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/152.0.0.0 Safari/537.36"
}

response = requests.get(url, headers)
response.encoding="utf-8"

soup = BeautifulSoup(response.text, "html.parser")

articles = soup.select("ul.list_newsheadline2 > li > a.item_newsheadline2")

for article in articles[0:9] :
    newsarticle_tag = article.select_one("div.cont_thumb strong.tit_txt")
    news_title = newsarticle_tag.get_text()
    print(news_title)
```
