```python
# type() : 함수
# count() : 메서드
# 함수는 앞에 어떤 것의 도움을 받지 않고 독립적으로 사용가능!
# 메서드는 사전적의미 방법, 거의 대부분 앞에 어떤 것 (= 객체)의 도움을 받아야함!
# 객체뒤에는 .(따옴표)필수로 붙여야함!

print(article.count("Korea")) # 인자값으로 입력된 문자열이 몇 번 사용되었는가? (대.소문자 구분함)
print(len(article)) # 전체 문자열 길이 갯수 (빈칸, 특수문자 모두포함)
print(article.find("s")) # 전체 문자열 중 몇번째 인텍스 = (순번)값으로 해당 인자값이 사용되었는지 확인! (0번부터 카운트시작, 빈칸, 특수문자 모두포함 / 해당사항이 없으면 -1)
```

```python
index = article.find("s")

while index != -1 :
    print(index)
    index = article.find("s", index + 1)
```

```python
float01 = 0.087
# type(float01)
print(format(float01, ".2f"))

# 바닷가 부유물(장치)
# 파이썬 내장 함수 -> 수치형 데이터 (정수, 실수자료형) -> 소수점 실수자료형 (절사, 내림, 올림, 반올림)
# 수학 라이브러리 모듈 math -> 더 쉽고 빠르게 (절사, 내림, 올림, 반올림)
```

```python
print("I have a pen, I have an apple")
print("I have a %s, I have an %s" % ("pen", "apple"))
print("I have a %d, I have an %d" % (10, 100))
print("I have a %c, I have an %c" % ('p', 't'))
print("I have a %f, I have an %f" % (3.14, 0.87))

"""
%s : string => 문자열 => "desk"
%c : charater => 문자 => 'a', 'b'
%d : decimal = int
%f : float
"""
```
