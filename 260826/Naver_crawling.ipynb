```python
import requests
import openpyxl
from bs4 import BeautifulSoup

import os
os.environ["JAVA_TOOL_OPTIONS"] = "--enable-na"

from konlpy.tag import Okt
# korean Natural Language processing in Python -> 형태소 분석 (문장 -> 어떤 단어들로 구성, 많이, 적게)
# Okt = Open Korean Text = 한국어를 기반으로한 형태소 분석기
from collections import Counter


excel_file = openpyxl.Workbook()
news_sheet = excel_file.active
news_sheet.title = "뉴스 수집"

news_sheet.append(["NO", "제목", "본문 미리보기", "링크"]) # field name = 필드명


query = input("검색하고싶은 키워드를 입력하세요")
url = f"https://search.naver.com/search.naver?where=news&ie=utf8&sm=nws_hty&query={query}"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36"
} # 딕션어리

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

news_contents = soup.select("div.sds-comps-vertical-layout.sds-comps-full-layout.vlFp8ERAn58mSKma")

all_bodies = []

for i, content in enumerate(news_contents[:10], 1) :
    title = content.select_one("a.fender-ui_228e3bd1.ZdTBe0dB_G0DGXFC")
    body = content.select_one("a.fender-ui_228e3bd1.DPWFguOj6E7T6a6U")
    
# titles = content.select("a.fender-ui_228e3bd1.ZdTBe0dB_G0DGXFC")
# bodies = content.select("a.fender-ui_228e3bd1.DPWFguOj6E7T6a6U")

    title_text = title.text.strip()
    link = title.get("href")
    body_text = body.text.strip()

    all_bodies.append(body_text)

    news_sheet.append([i, title_text, body_text, link])

# 문자열 -> 리스트 : split
# 리스트 -> 문자열 : join

full_text = " ".join(all_bodies)

okt = Okt()
nouns = okt.nouns(full_text)

# 불용어 설정 - 1 (내가 불필요)
stopwords = ["이번", "츄이", "대훈", "화의", "추이", "임명", "퇴사", "뉴스", "통신", "확인", "과연", "하이"]

# 불용어 설정 - 2 (단어 1개 / 카테고리에 따라 필요할수도 있음 EX)금,은,동)
filtered_nouns = [] #전역변수

for noun in nouns :
    if len(noun) >= 2 and noun not in stopwords :
        filtered_nouns.append(noun)

word_count = Counter(filtered_nouns)
top20 = word_count.most_common(20)

keyword_sheet = excel_file.create_sheet("키워드 형태소 분석")

keyword_sheet.append(["순위", "키워드", "빈도"])

for rank, (word, count) in enumerate(top20, 1) :
    keyword_sheet.append([rank, word, count])


news_sheet.column_dimensions["A"].width = 8
news_sheet.column_dimensions["B"].width = 50
news_sheet.column_dimensions["C"].width = 100
news_sheet.column_dimensions["D"].width = 80

keyword_sheet.column_dimensions["A"].width = 8
keyword_sheet.column_dimensions["B"].width = 30
keyword_sheet.column_dimensions["C"].width = 15

excel_file.save(f"{query}_네이버뉴스_형태소분석_데이터수집.xlsx")
excel_file.close()

print(f"{query}_네이버뉴스_형태소분석_데이터수집.xlsx 저장완료")


# for title in titles[:10] :
#     print(f"{i}. {title.text} {title.get("href")}")

# for body in bodies[:10] :
#     print(f"{i}. {body.text}")


# soup -> select : 인자값과 동일한 값이기만 하면 몇개든지 찾아와!
# 뉴스 기사 제목 따로 + 뉴스 본문 따로
# 뉴스 기사 & 본문 모두 가지고 있는 요소를 한 번만 찾아오면 찾아온 이후에 내부에서 각 요소를 조회
```
