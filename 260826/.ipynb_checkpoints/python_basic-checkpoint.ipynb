1. 프로그래밍 언어 역사
- 절차지향 프로그래밍 언어 (C언어 : 대부분의 프로그래밍 언어의 조상격)
- 절차지향 > 컴퓨터 이해하기 편하도록 > 모든 실행값 문법의 절차에 반드시 맞춰서 작성
- 객체지향 프로그래밍 언어 (Python, Javascript, Java) -> Class -> Instance
- 절차지향의 좋은점 수용 / 융통성이 없는 문법 배제
- 절차 -> 클래스 (Class -> 붕어빵 틀)
- 반죽 // 모양 틀 기계 // 성형하지 않아도X // 100, 1000

```python
print(type(int02), type(int03))
```

```python
# Python > crawling (무신사) > 베스트셀러 상품 tpo3 좋아요 및 리뷰수 평균 값 계산 (카테고리별)

# 20 > 60개 상품 좋아요, 리뷰수 > 평균 2,457 > 크롤링 > 정수 값 X > (문자 + 문자 + 문자) / 3
```

```python
# 1) 정수 = 숫자
# 사용할 수 있는 속성, 메서드, 연산자 상이함 = 다르다
# 속성 (변수 안에 있는 값의 정보를 추출할 때 사용할 수 있는 요소 = somthing = element)
# 메서드 (변수 안에 있는 값을 활용해서 기능 구현 = method)
```

```python
# 2) 정수라는 형태의 값이 변수에 할당 된다면, 가장 많이 사용하는 문법 -> 연산자!!!
# 사칙연산 (+, -, *, /) + @(나머지연산, 제곱근, 몫)

intA = 7
intB = 3

print(intA + intB)
print(intA - intB)
print(intA * intB)
print(intA / intB)
```

```python
print(intA // intB) # 7 /3 = 몫
print(intA % intB) # 7 % 3 = 나머지 값 => 값이 짝수, 홀수인지
print(intA ** intB) # 7^3
```

```python
# float > int(정수) 타입에서 사용가능한 모든 연산자 다 사용가능!
```

```python
# boolean >  True, False > 명제 > 조건문, 반복문

print(intA > intB) # 비교연산자
print(intA < intB)
print(intA == intB) # 변수선언, 할당 -> = : 할당의 의미
print(intA != intB) # ! -> 부정연산자 = Not
print(intA >= intB)
print(intA <= intB)
```

형변환 함수
- int() : 정수형태로 변환
- float() : 실수형태로 변환
- str() : 문자열로 변환
- bool() : 논리형으로 변환

```python
test01 = float(7)
test02 = str(7)
test03 = bool(0)
test04 = bool(7)

print(test01, test02, type(test02), test03, test04)
```
