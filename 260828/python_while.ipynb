```python
data = int(input("구구단 몇 단을 제작하시겠습니까? ex. 2단 -> 2"))
# 조건 : 2~9단 까지만 구구단을 생성해라! 10 ~ X
if data >= 2 and data <=9 :
    for digit in range(1, 10) :
        print(f"{data} x {digit} = {data * digit}")
else :
    print("실행할 수 없습니다!")
```
